```python
import pandas as pd
import numpy as np
import random
```

# random data generation

```python
data = {
    "total_matches": np.random.randint(20,100,100),
    "centuries": np.random.randint(20,100,100),
    "half_centuries": np.random.randint(20,100,100),
    "highest_score": np.random.randint(20,100,100),
    "total_runs": np.random.randint(20,100,100),
    "strike_rate": np.random.randint(20,100,100),
}
df = pd.DataFrame(data)
```

## mean

```python
original_mean = df.mean()
print(original_mean)
```

## srs

```python
srs = df.sample(3,random_state = 1)
srs_mean = srs.mean()
srs_mean
srs_difference = original_mean - srs_mean
srs_difference
```

# ssm

```python
ssm = df[df["total_matches"]>25].sample(5,random_state = 1)
ssm
ssm_mean = ssm.mean()
ssm_mean
ssm_difference = ssm_mean - original_mean
ssm_difference
```

### cluster sampling

```python
#ddivide data into groups and selects the clusters
#eg divides the schools into clusters and randomply selects any 2 of them amd perform actions
clusters = ['a','b','c','d']
df["div"] = np.random.choice(clusters, size=100)
clust_data = np.random.choice(clusters,2)
print(clust_data)
sample_data = df[df["div"].isin(clust_data)]
print(sample_data.head())
```
