### variance

```python
import pandas as pd
import numpy as np
df = pd.read_csv("Netflix Userbase.csv")

df.head()



```

#### checking is null

```python
print(df.isnull().sum())
print("hii")
print(df.duplicated().sum())
```

### mean

```python
print(df["Monthly Revenue"].head())
sum = df["Monthly Revenue"].sum()
print(sum)
mean = sum/len(df["Monthly Revenue"])
print(mean)
```

### median

```python
sorted_values = df["Monthly Revenue"].sort_values().values
n = len(sorted_values)
if len(sorted_values) %2 == 0:
    print((sorted_values[n//2 -1]+sorted_values[n//2])/2)
else:
    print(sorted_values[n//2])
np.median(df["Monthly Revenue"])
```

### mode

```python
count = df["Monthly Revenue"].value_counts()# we can aslo use groupby(df[monthly_revenue]).size() size counts null values also nut count not counts null values
print(count.index[0])
```

### variance


```python
deviation = (df["Monthly Revenue"]- mean) ** 2
variance = deviation.sum()/len(df["Monthly Revenue"])
print(variance)
np.var(df["Monthly Revenue"])
```

### standard deviation

```python
standard_deviation = np.sqrt(variance)
print(standard_deviation)
print(np.std(df["Monthly Revenue"]))
```
