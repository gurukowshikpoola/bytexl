### data

```python
import matplotlib.pyplot as plt 
import pandas as pd 
df={"Studs":["Rahul","Rohit","Hemanth","Rishita","Raga"],
    "Marks":[85,78,91,79,86],
    "Age":[20,21,23,22,28]}

df=pd.DataFrame(df)
df
```

### normal grpah
attribute:color,width,edgecolor,linewidth

```python
plt.figure(figsize=(7,4),facecolor="blue") # it creates a canvas on which will show tha graph 
plt.bar(df["Studs"],df["Marks"])
plt.title("Data Viz")
plt.show()
```

## plot
plot attributes: color,linestyle:it has -,--,:(dotted),-.
                linewidth,marker=o,s(squares),^,x,+
                marker size
                

```python
import matplotlib.pyplot as plt
import pandas as pd
months = ["jan","feb","mar","apr","may"]
sales = [20000,23000,16000,30000,10000]
plt.figure(figsize=(7, 4),facecolor="white")
plt.plot(months,sales,marker = "*",markersize=10,label="salesdata", lw = 2,ls="--")
plt.title("data viz")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

### scatterplot
color,s(pointer size),marker,alpha

```python
df2={
    "studyhours":[1,2,3,4,5,2.5,3.5,4.5],
    "marksscored":[50,55,65,70,85,60,68,75]
}
plt.figure(figsize=(7,4),facecolor="white")
plt.scatter(df2["studyhours"],df2["marksscored"],s=50,c="red",marker="o",lw=2,alpha=0.7)
plt.title("sctter plot")
plt.show()
```

### pie
colors=["r","b",""]used to color measurement
startandle = rotate the pie by degrees
autopct = shows percentages
labels = names of parts
explod = separate slices
shadow = shadow effrcet(shadow = true )# to get shadow


```python
df2={
    "studyhours":[1,2,3,4,5,2.5,3.5,4.5],
    "marksscored":[50,55,65,70,85,60,68,75]
}
import matplotlib.pyplot as plt 
import pandas as pd 
plt.figure(figsize=(7,4),facecolor="white")
plt.plot(df2["studyhours"],df2["marksscored"],marker="*",markersize=10,label="Sales Data ",lw=1,ls="--")
plt.title("Data Viz ")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True,alpha=0.2)
plt.show()
```

# histogram

```python
import matplotlib.pyplot as plt
import pandas as np
ages = [15,16,17,16,15,18,16,15,18,16,15,18,18,16,16]
plt.figure(figsize=(8,4),facecolor="r")
plt.hist(ages,color="blue",edgecolor = "yellow",lw = 2,alpha = 0.96)
plt.title("histogram")
plt.xlabel("ages in years")
plt.ylabel("no.of students")
plt.show()
```
