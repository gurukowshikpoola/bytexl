## plot

```python
import matplotlib.pyplot as plt
import pandas as pd
months = ["jan","feb","mar","apr","may"]
sales = ["20000","23000","16000","30000","10000"]
plt.figure(figsize=(7, 4),facecolor="white")
plt.plot(months,sales,marker = "*",markersize=10,label="salesdata", lw = 2,ls="--")
plt.title("data viz")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```
